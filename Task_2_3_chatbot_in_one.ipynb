```python
from google.colab import userdata
API_KEY = userdata.get("TEST_API")
from google import genai
from google.genai import types
client=genai.Client(api_key=API_KEY)
prompt="What is the Python language?"
model1="gemini-3.6-flash"
```

```python
# chat_bot1
response_1=client.models.generate_content(
    model=model1,
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=1.4,
        top_p=0.5,
        system_instruction="""You are an experienced programming instructor.
Explain Python in a simple and beginner-friendly way.
Use short sentences and avoid technical jargon.
Give one simple real-world example.
Limit the answer to about 100 words."""
    )
)
```

```python
# chat_bot2
response_2=client.models.generate_content(
    model=model1,
    contents=prompt,
    config=types.GenerateContentConfig(
        temperature=1.4,
        top_p=0.5,
        system_instruction="""You are a senior software engineer.
Explain Python from a technical perspective.
Discuss its interpreted nature, dynamic typing, garbage collection, ecosystem, and common use cases.
Assume the reader has programming experience.
Limit the answer to about 200 words."""
    )
)
```

```python
# chat_bot3  (Judge)
judge_prompt = f"""
Question:
{prompt}
Response 1:
{response_1.text}
Response 2:
{response_2.text}
Compare these two responses.
"""
judge = client.models.generate_content(
    model=model1,
    contents=judge_prompt,
    config=types.GenerateContentConfig(
        system_instruction="""
You are an expert AI evaluator.
Evaluate both responses for:
- Accuracy
- Completeness
- Clarity
- Relevance
- Technical correctness
Give each response a score from 1 to 10.
Choose the better response and explain why.
"""
    )
)
print(judge.text)
```

```python
print(response_1.usage_metadata.prompt_token_count)
print(response_2.usage_metadata.prompt_token_count)
print(judge.usage_metadata.prompt_token_count)
```

```python
print(response_1.usage_metadata.candidates_token_count)
print(response_2.usage_metadata.candidates_token_count)
print(judge.usage_metadata.candidates_token_count)
```
